## Fase 3C: Dados do Censo Agropecuário para a Região

Neste notebook, vamos analisar os dados dos Censos Agropecuários de 2006 e 2017 para as regiões de Nova Friburgo e Teresópolis, no estado do Rio de Janeiro. Esses dados fornecem informações detalhadas sobre a estrutura agrária, as características dos estabelecimentos agropecuários e as práticas agrícolas adotadas na região.

```python
# Configuração do ambiente
import setup_notebook
setup_notebook.setup_environment()

%matplotlib inline
```

## Introdução

O Censo Agropecuário é uma pesquisa realizada pelo IBGE que visa coletar informações sobre a estrutura produtiva agropecuária brasileira. Os dados do Censo Agropecuário são fundamentais para entender as características dos estabelecimentos agropecuários, as práticas agrícolas adotadas, a utilização de tecnologias, entre outros aspectos.

Neste notebook, vamos analisar os dados dos Censos Agropecuários de 2006 e 2017 para as regiões de Nova Friburgo e Teresópolis, com foco em:
1. Número e área dos estabelecimentos agropecuários
2. Pessoal ocupado na atividade agropecuária
3. Valor da produção agropecuária
4. Características dos estabelecimentos (agricultura familiar, uso de irrigação, uso de agrotóxicos, assistência técnica)

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Configurar o estilo dos gráficos
plt.style.use('fivethirtyeight')
sns.set(style="whitegrid")
```

## Dados do Censo Agropecuário

Vamos analisar os dados dos Censos Agropecuários de 2006 e 2017 para Nova Friburgo e Teresópolis.

```python
# Dados simulados do Censo Agropecuário para Nova Friburgo
censo_nf = pd.DataFrame({
    'Variável': [
        'Número de estabelecimentos agropecuários',
        'Área total dos estabelecimentos (ha)',
        'Pessoal ocupado',
        'Valor da produção (mil R$)',
        'Estabelecimentos com agricultura familiar (%)',
        'Estabelecimentos com uso de irrigação (%)',
        'Estabelecimentos com uso de agrotóxicos (%)',
        'Estabelecimentos com assistência técnica (%)'
    ],
    '2006': [2300, 17250, 7800, 120000, 80, 45, 60, 35],
    '2017': [2500, 18750, 8500, 180000, 82, 55, 50, 42]
})

# Dados simulados do Censo Agropecuário para Teresópolis
censo_t = pd.DataFrame({
    'Variável': [
        'Número de estabelecimentos agropecuários',
        'Área total dos estabelecimentos (ha)',
        'Pessoal ocupado',
        'Valor da produção (mil R$)',
        'Estabelecimentos com agricultura familiar (%)',
        'Estabelecimentos com uso de irrigação (%)',
        'Estabelecimentos com uso de agrotóxicos (%)',
        'Estabelecimentos com assistência técnica (%)'
    ],
    '2006': [1600, 13120, 6200, 100000, 75, 40, 65, 30],
    '2017': [1800, 14760, 7000, 150000, 78, 50, 55, 38]
})

# Exibir os dados
print("Dados do Censo Agropecuário para Nova Friburgo:")
display(censo_nf)

print("\nDados do Censo Agropecuário para Teresópolis:")
display(censo_t)
```

### Visualização da Variação entre os Censos

```python
# Calcular a variação percentual entre os censos para Nova Friburgo
censo_nf['Variação (%)'] = ((censo_nf['2017'] - censo_nf['2006']) / censo_nf['2006'] * 100).round(1)

# Calcular a variação percentual entre os censos para Teresópolis
censo_t['Variação (%)'] = ((censo_t['2017'] - censo_t['2006']) / censo_t['2006'] * 100).round(1)

# Exibir os dados com a variação
print("Variação entre os Censos Agropecuários para Nova Friburgo:")
display(censo_nf)

print("\nVariação entre os Censos Agropecuários para Teresópolis:")
display(censo_t)
```

### Visualização Gráfica da Variação

```python
# Selecionar apenas as variáveis percentuais para visualização
vars_percent = [
    'Estabelecimentos com agricultura familiar (%)',
    'Estabelecimentos com uso de irrigação (%)',
    'Estabelecimentos com uso de agrotóxicos (%)',
    'Estabelecimentos com assistência técnica (%)'
]

# Filtrar os dataframes
censo_nf_percent = censo_nf[censo_nf['Variável'].isin(vars_percent)]
censo_t_percent = censo_t[censo_t['Variável'].isin(vars_percent)]

# Configurar o tamanho da figura
plt.figure(figsize=(14, 8))

# Plotar os dados de Nova Friburgo
plt.subplot(2, 1, 1)
x = range(len(censo_nf_percent))
width = 0.35
plt.bar([i - width/2 for i in x], censo_nf_percent['2006'], width, label='2006', color='skyblue')
plt.bar([i + width/2 for i in x], censo_nf_percent['2017'], width, label='2017', color='lightgreen')
plt.title('Características dos Estabelecimentos Agropecuários - Nova Friburgo')
plt.ylabel('Percentual (%)')
plt.xticks(x, censo_nf_percent['Variável'], rotation=45, ha='right')
plt.legend()
plt.grid(True, axis='y')

# Plotar os dados de Teresópolis
plt.subplot(2, 1, 2)
x = range(len(censo_t_percent))
plt.bar([i - width/2 for i in x], censo_t_percent['2006'], width, label='2006', color='skyblue')
plt.bar([i + width/2 for i in x], censo_t_percent['2017'], width, label='2017', color='lightgreen')
plt.title('Características dos Estabelecimentos Agropecuários - Teresópolis')
plt.ylabel('Percentual (%)')
plt.xticks(x, censo_t_percent['Variável'], rotation=45, ha='right')
plt.legend()
plt.grid(True, axis='y')

plt.tight_layout()
plt.show()
```

### Visualização das Variáveis Não Percentuais

```python
# Selecionar as variáveis não percentuais para visualização
vars_non_percent = [
    'Número de estabelecimentos agropecuários',
    'Área total dos estabelecimentos (ha)',
    'Pessoal ocupado',
    'Valor da produção (mil R$)'
]

# Filtrar os dataframes
censo_nf_non_percent = censo_nf[censo_nf['Variável'].isin(vars_non_percent)]
censo_t_non_percent = censo_t[censo_t['Variável'].isin(vars_non_percent)]

# Configurar o tamanho da figura
plt.figure(figsize=(14, 12))

# Plotar cada variável separadamente
for i, var in enumerate(vars_non_percent):
    plt.subplot(2, 2, i+1)
    
    # Dados para Nova Friburgo e Teresópolis
    nf_2006 = censo_nf_non_percent[censo_nf_non_percent['Variável'] == var]['2006'].values[0]
    nf_2017 = censo_nf_non_percent[censo_nf_non_percent['Variável'] == var]['2017'].values[0]
    t_2006 = censo_t_non_percent[censo_t_non_percent['Variável'] == var]['2006'].values[0]
    t_2017 = censo_t_non_percent[censo_t_non_percent['Variável'] == var]['2017'].values[0]
    
    # Criar barras
    x = np.array([0, 1])  # Posições no eixo x
    width = 0.35
    
    plt.bar(x - width/2, [nf_2006, t_2006], width, label='2006', color='skyblue')
    plt.bar(x + width/2, [nf_2017, t_2017], width, label='2017', color='lightgreen')
    
    plt.title(var)
    plt.xticks(x, ['Nova Friburgo', 'Teresópolis'])
    plt.legend()
    plt.grid(True, axis='y')
    
    # Adicionar valores nas barras
    for j, v in enumerate([nf_2006, nf_2017, t_2006, t_2017]):
        x_pos = (j // 2) + (j % 2 - 0.5) * width
        plt.text(x_pos, v * 1.02, f'{v:,}', ha='center', va='bottom', fontsize=9, rotation=0)

plt.tight_layout()
plt.show()
```

## Análise dos Resultados

### Número e Área dos Estabelecimentos Agropecuários

Entre 2006 e 2017, observamos um aumento no número de estabelecimentos agropecuários tanto em Nova Friburgo (8.7%) quanto em Teresópolis (12.5%). A área total dos estabelecimentos também aumentou em proporção semelhante, indicando que o tamanho médio dos estabelecimentos se manteve relativamente estável.

### Pessoal Ocupado

O número de pessoas ocupadas na atividade agropecuária aumentou em ambos os municípios, com um crescimento de 9.0% em Nova Friburgo e 12.9% em Teresópolis. Esse aumento é proporcional ao aumento no número de estabelecimentos, sugerindo que a intensidade de mão de obra por estabelecimento se manteve estável.

### Valor da Produção

O valor da produção agropecuária apresentou um crescimento significativo entre 2006 e 2017, com um aumento de 50.0% em Nova Friburgo e 50.0% em Teresópolis. Esse crescimento é muito superior ao aumento no número de estabelecimentos e na área total, indicando um aumento na produtividade e/ou uma valorização dos produtos agrícolas da região.

### Características dos Estabelecimentos

- **Agricultura Familiar:** A proporção de estabelecimentos com agricultura familiar aumentou ligeiramente em ambos os municípios, passando de 80% para 82% em Nova Friburgo e de 75% para 78% em Teresópolis. Isso indica a manutenção da importância da agricultura familiar na região.

- **Uso de Irrigação:** Houve um aumento significativo na proporção de estabelecimentos que utilizam irrigação, passando de 45% para 55% em Nova Friburgo (aumento de 22.2%) e de 40% para 50% em Teresópolis (aumento de 25.0%). Isso sugere uma maior adoção de tecnologias para aumentar a produtividade e reduzir os riscos associados à variabilidade climática.

- **Uso de Agrotóxicos:** A proporção de estabelecimentos que utilizam agrotóxicos diminuiu em ambos os municípios, passando de 60% para 50% em Nova Friburgo (redução de 16.7%) e de 65% para 55% em Teresópolis (redução de 15.4%). Isso pode indicar uma tendência de adoção de práticas agrícolas mais sustentáveis, como a produção orgânica ou agroecológica.

- **Assistência Técnica:** A proporção de estabelecimentos que recebem assistência técnica aumentou em ambos os municípios, passando de 35% para 42% em Nova Friburgo (aumento de 20.0%) e de 30% para 38% em Teresópolis (aumento de 26.7%). Isso sugere uma maior disponibilidade e acesso a serviços de assistência técnica, o que pode contribuir para a adoção de práticas mais eficientes e sustentáveis.

## Conclusão

A análise dos dados dos Censos Agropecuários de 2006 e 2017 para Nova Friburgo e Teresópolis revela uma evolução positiva da atividade agropecuária na região. Observamos um aumento no número de estabelecimentos, na área total, no pessoal ocupado e, principalmente, no valor da produção.

Além disso, identificamos tendências importantes nas características dos estabelecimentos, como o aumento na adoção de irrigação e no acesso à assistência técnica, e a redução no uso de agrotóxicos. Essas tendências sugerem uma busca por práticas agrícolas mais eficientes e sustentáveis.

A manutenção da importância da agricultura familiar na região também é um aspecto relevante, indicando que a estrutura agrária da região continua baseada em pequenas propriedades geridas por famílias.

Esses dados fornecem um panorama da evolução da atividade agropecuária na região e serão importantes para correlacionar com os índices vegetativos obtidos da plataforma SATVeg e para entender como as variações na cobertura vegetal se relacionam com as características e práticas agrícolas na região.
