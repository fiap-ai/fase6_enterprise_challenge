## Introdução

Na Parte 1 deste notebook, realizamos a definição das variáveis-chave para o modelo e uma análise inicial da relação entre os índices NDVI/EVI e a produtividade agrícola. Agora, vamos continuar a análise com foco nas seguintes tarefas:

1. Identificação dos períodos críticos de crescimento da cultura

3. Extração de features temporais (médias, máximos, mínimos, tendências)
4. Seleção das features mais relevantes para o modelo de IA

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime
import os
from sklearn.feature_selection import SelectKBest, f_regression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.decomposition import PCA
from sklearn.model_selection import train_test_split
from scipy import stats

# Configurar o estilo dos gráficos
plt.style.use('fivethirtyeight')
sns.set(style="whitegrid")
```

## 1. Carregamento dos Dados

Vamos carregar os dados pré-processados e os resultados da Parte 1.

```python
# Carregar os dados de NDVI/EVI
df_ndvi_nf = pd.read_csv('../../assets/ndvi_mensal_nova_friburgo.csv')
df_ndvi_t = pd.read_csv('../../assets/ndvi_mensal_teresopolis.csv')

# Carregar os dados de produtividade agrícola
df_prod_combinados = pd.read_csv('../../assets/dados_produtividade_combinados.csv')

# Filtrar os dados de produtividade para o ano de 2017
df_prod_2017 = df_prod_combinados[df_prod_combinados['Ano'] == 2017]

# Agrupar os dados de produtividade por município e tipo de lavoura
df_prod_grouped = df_prod_2017.groupby(['Município', 'Tipo de Lavoura'])['Produtividade (t/ha)'].mean().reset_index()

# Calcular a média de produtividade por município
df_prod_mean = df_prod_grouped.groupby('Município')['Produtividade (t/ha)'].mean().reset_index()

# Exibir os dados de produtividade
print("Produtividade média por município em 2017:")
print(df_prod_mean)
```

```python
# Converter a coluna de data para o formato datetime
df_ndvi_nf['Data'] = pd.to_datetime(df_ndvi_nf['Ano'].astype(str) + '-' + df_ndvi_nf['Mês'].astype(str) + '-01')
df_ndvi_t['Data'] = pd.to_datetime(df_ndvi_t['Ano'].astype(str) + '-' + df_ndvi_t['Mês'].astype(str) + '-01')

# Filtrar os dados para o ano de 2017
df_ndvi_nf_2017 = df_ndvi_nf[df_ndvi_nf['Ano'] == 2017].copy()
df_ndvi_t_2017 = df_ndvi_t[df_ndvi_t['Ano'] == 2017].copy()

# Adicionar coluna de município
df_ndvi_nf_2017['Município'] = 'Nova Friburgo'
df_ndvi_t_2017['Município'] = 'Teresópolis'

# Concatenar os dataframes
df_ndvi_2017 = pd.concat([df_ndvi_nf_2017, df_ndvi_t_2017])

# Exibir as primeiras linhas do dataframe combinado
print("Primeiras linhas do dataframe combinado para 2017:")
print(df_ndvi_2017[['Município', 'Ano', 'Mês', 'Data', 'EVI_mean', 'SG_mean']].head())
```

## 2. Identificação dos Períodos Críticos de Crescimento da Cultura

Vamos analisar a relação entre os índices NDVI/EVI em diferentes meses e a produtividade agrícola para identificar os períodos críticos de crescimento da cultura.

```python
# Pivotar o dataframe para ter uma linha por município e colunas para cada mês
df_monthly_pivot = df_ndvi_2017.pivot_table(
    index='Município',
    columns='Mês',
    values=['EVI_mean', 'SG_mean']
).reset_index()

# Renomear as colunas
df_monthly_pivot.columns = ['Município'] + [
    f"{col[0]}_mes_{col[1]}" for col in df_monthly_pivot.columns.values[1:]
]

# Mesclar com os dados de produtividade
df_monthly_features = pd.merge(df_monthly_pivot, df_prod_mean, on='Município')

# Exibir o dataframe com as features mensais e a produtividade
print("Dataframe com features mensais e produtividade para 2017:")
print(df_monthly_features)
```

```python
# Vamos calcular a correlação entre os índices mensais e a produtividade
# Primeiro, vamos transpor o dataframe para facilitar o cálculo
df_monthly_features_t = df_monthly_features.set_index('Município').T

# Calcular a correlação entre os índices e a produtividade
corr_nf = df_monthly_features_t.corrwith(df_monthly_features_t.loc['Produtividade (t/ha)', 'Nova Friburgo'])
corr_t = df_monthly_features_t.corrwith(df_monthly_features_t.loc['Produtividade (t/ha)', 'Teresópolis'])

# Criar um dataframe com as correlações
df_corr = pd.DataFrame({
    'Correlação Nova Friburgo': corr_nf,
    'Correlação Teresópolis': corr_t
})

# Remover a linha da produtividade
df_corr = df_corr.drop('Produtividade (t/ha)')

# Ordenar por correlação absoluta média
df_corr['Correlação Média Absoluta'] = df_corr.abs().mean(axis=1)
df_corr = df_corr.sort_values('Correlação Média Absoluta', ascending=False)

# Exibir as correlações
print("Correlação entre os índices mensais e a produtividade:")
print(df_corr.head(10))
```

```python
# Vamos visualizar as correlações em um gráfico de barras
plt.figure(figsize=(12, 8))

# Selecionar as 10 features com maior correlação absoluta média
top_features = df_corr.head(10).index

# Criar um dataframe com as correlações das top features
df_corr_top = df_corr.loc[top_features, ['Correlação Nova Friburgo', 'Correlação Teresópolis']]

# Plotar o gráfico de barras
df_corr_top.plot(kind='bar', figsize=(12, 8))
plt.title('Correlação entre os Índices Mensais e a Produtividade')
plt.xlabel('Índice Mensal')
plt.ylabel('Correlação')
plt.xticks(rotation=45, ha='right')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Vamos extrair o mês de cada feature
df_corr['Mês'] = df_corr.index.str.extract(r'mes_(\d+)').astype(float)

# Filtrar apenas as features de EVI
df_corr_evi = df_corr[df_corr.index.str.startswith('EVI_mean')]

# Ordenar por mês
df_corr_evi = df_corr_evi.sort_values('Mês')

# Plotar a correlação por mês para EVI
plt.figure(figsize=(12, 6))
plt.plot(df_corr_evi['Mês'], df_corr_evi['Correlação Nova Friburgo'], 'g-', marker='o', 
         linewidth=2, label='Nova Friburgo')
plt.plot(df_corr_evi['Mês'], df_corr_evi['Correlação Teresópolis'], 'b-', marker='o', 
         linewidth=2, label='Teresópolis')
plt.title('Correlação entre EVI Mensal e Produtividade')
plt.xlabel('Mês')
plt.ylabel('Correlação')
plt.xticks(range(1, 13), ['Jan', 'Fev', 'Mar', 'Abr', 'Mai', 'Jun', 'Jul', 'Ago', 'Set', 'Out', 'Nov', 'Dez'])
plt.axhline(y=0, color='r', linestyle='-', alpha=0.3)
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Filtrar apenas as features de SG
df_corr_sg = df_corr[df_corr.index.str.startswith('SG_mean')]

# Ordenar por mês
df_corr_sg = df_corr_sg.sort_values('Mês')

# Plotar a correlação por mês para SG
plt.figure(figsize=(12, 6))
plt.plot(df_corr_sg['Mês'], df_corr_sg['Correlação Nova Friburgo'], 'g-', marker='o', 
         linewidth=2, label='Nova Friburgo')
plt.plot(df_corr_sg['Mês'], df_corr_sg['Correlação Teresópolis'], 'b-', marker='o', 
         linewidth=2, label='Teresópolis')
plt.title('Correlação entre Savitzky-Golay Mensal e Produtividade')
plt.xlabel('Mês')
plt.ylabel('Correlação')
plt.xticks(range(1, 13), ['Jan', 'Fev', 'Mar', 'Abr', 'Mai', 'Jun', 'Jul', 'Ago', 'Set', 'Out', 'Nov', 'Dez'])
plt.axhline(y=0, color='r', linestyle='-', alpha=0.3)
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

## 3. Extração de Features Temporais

Com base na análise anterior, vamos extrair features temporais relevantes para o modelo de IA.

```python
# Vamos selecionar as features mais relevantes com base na análise de correlação
# Vamos considerar os meses com maior correlação absoluta média

# Selecionar as 5 features com maior correlação absoluta média
top_5_features = df_corr.head(5).index

# Criar um dataframe com as features selecionadas
df_selected_features = df_monthly_features[['Município', 'Produtividade (t/ha)'] + list(top_5_features)]

# Exibir o dataframe com as features selecionadas
print("Dataframe com as features selecionadas:")
print(df_selected_features)
```

```python
# Vamos criar features adicionais baseadas nos períodos críticos identificados
# Vamos calcular a média, mínimo, máximo e desvio padrão do EVI e SG para os períodos críticos

# Identificar os meses críticos com base na análise de correlação
# Vamos considerar os meses com maior correlação absoluta média
critical_months = df_corr_evi.sort_values('Correlação Média Absoluta', ascending=False).head(3)['Mês'].astype(int).tolist()

print(f"Meses críticos identificados: {critical_months}")

# Calcular features para os períodos críticos
critical_features = {}

for municipio in ['Nova Friburgo', 'Teresópolis']:
    # Filtrar os dados para o município e os meses críticos
    df_critical = df_ndvi_2017[(df_ndvi_2017['Município'] == municipio) & 
                              (df_ndvi_2017['Mês'].isin(critical_months))]
    
    # Calcular estatísticas para os períodos críticos
    critical_features[municipio] = {
        'EVI_critical_mean': df_critical['EVI_mean'].mean(),
        'EVI_critical_min': df_critical['EVI_mean'].min(),
        'EVI_critical_max': df_critical['EVI_mean'].max(),
        'EVI_critical_std': df_critical['EVI_mean'].std(),
        'SG_critical_mean': df_critical['SG_mean'].mean(),
        'SG_critical_min': df_critical['SG_mean'].min(),
        'SG_critical_max': df_critical['SG_mean'].max(),
        'SG_critical_std': df_critical['SG_mean'].std()
    }

# Criar um dataframe com as features dos períodos críticos
df_critical_features = pd.DataFrame.from_dict(critical_features, orient='index').reset_index()
df_critical_features.rename(columns={'index': 'Município'}, inplace=True)

# Exibir o dataframe com as features dos períodos críticos
print("\nDataframe com as features dos períodos críticos:")
print(df_critical_features)
```

```python
# Mesclar com as features selecionadas
df_final_features = pd.merge(df_selected_features, df_critical_features, on='Município')

# Exibir o dataframe com as features finais
print("Dataframe com as features finais:")
print(df_final_features)
```

## 4. Seleção das Features Mais Relevantes

Vamos selecionar as features mais relevantes para o modelo de IA.

```python
# Vamos calcular a correlação entre todas as features e a produtividade
# Primeiro, vamos remover a coluna de município
df_corr_final = df_final_features.drop(columns=['Município']).corr()['Produtividade (t/ha)'].sort_values(ascending=False)

# Exibir as correlações
print("Correlação entre as features finais e a produtividade:")
print(df_corr_final)
```

```python
# Vamos visualizar as correlações em um gráfico de barras
plt.figure(figsize=(12, 8))
df_corr_final.drop('Produtividade (t/ha)').plot(kind='bar', figsize=(12, 8))
plt.title('Correlação entre as Features Finais e a Produtividade')
plt.xlabel('Feature')
plt.ylabel('Correlação')
plt.xticks(rotation=45, ha='right')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Vamos selecionar as features mais relevantes com base na correlação
# Vamos considerar as features com correlação absoluta acima de 0.5
relevant_features = df_corr_final.drop('Produtividade (t/ha)').abs().sort_values(ascending=False)
relevant_features = relevant_features[relevant_features > 0.5].index.tolist()

# Exibir as features relevantes
print(f"Features relevantes selecionadas: {relevant_features}")

# Criar um dataframe com as features relevantes
df_relevant_features = df_final_features[['Município', 'Produtividade (t/ha)'] + relevant_features]

# Exibir o dataframe com as features relevantes
print("\nDataframe com as features relevantes:")
print(df_relevant_features)
```

## 5. Exportação das Features para o Modelo de IA

Vamos exportar as features selecionadas para serem utilizadas no desenvolvimento do modelo de IA.

```python
# Criar diretório para os dados processados
os.makedirs('../../sprint2/assets', exist_ok=True)

# Exportar as features finais
df_final_features.to_csv('../../sprint2/assets/features_finais.csv', index=False)

# Exportar as features relevantes
df_relevant_features.to_csv('../../sprint2/assets/features_relevantes.csv', index=False)

# Exportar a lista de meses críticos
pd.DataFrame({'Mês Crítico': critical_months}).to_csv('../../sprint2/assets/meses_criticos.csv', index=False)

print("Features exportadas com sucesso!")
```
